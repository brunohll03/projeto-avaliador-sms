Pré-processamento do texto

converter para letras minúsculas;
remover espaços extras;
tratar valores ausentes.

```python
# Importa a biblioteca re para trabalhar com expressões regulares.
import re
```

```python
# Define uma função responsável pela normalização básica de uma mensagem.
def normalizar_mensagem(mensagem):
    
    # Verifica se o conteúdo recebido é nulo.
    if pd.isna(mensagem):
        
        # Retorna uma string vazia caso a mensagem esteja ausente.
        return ""
    
    # Converte o conteúdo da mensagem para texto.
    mensagem = str(mensagem)
    
    # Converte todas as letras da mensagem para minúsculas.
    mensagem = mensagem.lower()
    
    # Substitui sequências de espaços, quebras de linha e tabulações por um único espaço.
    mensagem = re.sub(r"\s+", " ", mensagem)
    
    # Remove espaços desnecessários existentes no início e no final da mensagem.
    mensagem = mensagem.strip()
    
    # Retorna a mensagem normalizada.
    return mensagem
```

Aplicando:

Próximo passo

Agora que o texto está normalizado, o próximo passo será transformar as mensagens em números utilizando TF-IDF.

Mas existe um ponto crucial que quero que você entenda antes de fazermos isso:

O TF-IDF será ajustado (fit) somente nas mensagens de treinamento.

Depois faremos apenas transform nas mensagens de teste.

Isso é exatamente para evitar data leakage.

Transformação do texto com TF-IDF

```python
# Importa o transformador TF-IDF da biblioteca Scikit-learn.
from sklearn.feature_extraction.text import TfidfVectorizer
```

Criar o vetor TF-IDF

```python
# Cria o vetor TF-IDF que será utilizado para transformar as mensagens em valores numéricos.
vetorizador_tfidf = TfidfVectorizer()
```

Fazer o fit_transform somente no treinamento

fit
 ↓
aprende o vocabulário do treinamento

transform
 ↓
converte as mensagens em números

```python
# Aprende o vocabulário e os pesos TF-IDF utilizando somente as mensagens de treinamento.
mensagens_treino_tfidf = vetorizador_tfidf.fit_transform(mensagens_treino_normalizadas)
```

Transformar o conjunto de teste

```python
# Utiliza o vocabulário aprendido no treinamento para transformar as mensagens de teste.
mensagens_teste_tfidf = vetorizador_tfidf.transform(mensagens_teste_normalizadas)
```

Verificar o tamanho das matrizes

```python
# Exibe a quantidade de mensagens e características TF-IDF existentes no conjunto de treinamento.
print("Formato TF-IDF treinamento:", mensagens_treino_tfidf.shape)

# Exibe a quantidade de mensagens e características TF-IDF existentes no conjunto de teste.
print("Formato TF-IDF teste:", mensagens_teste_tfidf.shape)
```

Guardar o vetor para uso futuro

```python
# Importa o módulo joblib para salvar objetos Python no disco.
import joblib
```

Combinar TF-IDF com os atributos estruturados

```python
# Importa a função hstack para combinar matrizes horizontalmente.
from scipy.sparse import hstack

# Converte os atributos estruturados de treinamento para uma matriz NumPy.
atributos_treino_array = X_atributos_treino.to_numpy()

# Converte os atributos estruturados de teste para uma matriz NumPy.
atributos_teste_array = X_atributos_teste.to_numpy()

# Combina as características TF-IDF com os atributos estruturados do treinamento.
X_treino_final = hstack([mensagens_treino_tfidf, atributos_treino_array])

# Combina as características TF-IDF com os atributos estruturados do teste.
X_teste_final = hstack([mensagens_teste_tfidf, atributos_teste_array])
```

Guardar os nomes das características

```python
# Obtém a lista com os nomes das características estruturadas.
nomes_atributos = X_atributos.columns.tolist()

# Obtém a lista com os termos que foram aprendidos pelo TF-IDF.
nomes_tfidf = vetorizador_tfidf.get_feature_names_out().tolist()

# Combina os nomes das características de texto com os nomes das características estruturadas.
nomes_features = nomes_tfidf + nomes_atributos

# Exibe a quantidade total de nomes de características.
print("Total de features:", len(nomes_features))
```

Salvar tudo que foi preparado

```python
# Importa a biblioteca joblib para salvar objetos Python em arquivos.
import joblib
```

```python
# Cria um dicionário vazio para armazenar os objetos preparados.
dados_preparados = {}

# Armazena as características finais utilizadas no treinamento.
dados_preparados["X_treino_final"] = X_treino_final

# Armazena as características finais utilizadas no teste.
dados_preparados["X_teste_final"] = X_teste_final

# Armazena os valores alvo utilizados no treinamento.
dados_preparados["y_treino"] = y_treino

# Armazena os valores alvo utilizados no teste.
dados_preparados["y_teste"] = y_teste

# Armazena o vetorizador TF-IDF treinado.
dados_preparados["vetorizador_tfidf"] = vetorizador_tfidf

# Armazena os nomes de todas as características utilizadas pelo modelo.
dados_preparados["nomes_features"] = nomes_features

# Salva todos os dados preparados dentro da pasta de dados processados.
joblib.dump(dados_preparados, "../data/processed/dados_preparados.joblib")
```
