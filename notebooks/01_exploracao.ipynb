Importando o Pandas

```python
import pandas as pd  # Importa a biblioteca Pandas para manipulação e análise de dados
```

Definir o caminho do dataset

```python
caminho_dataset = "../data/raw/DATASET - Dataset_limpo_V1.csv"  # Define o caminho relativo até o arquivo CSV
```

Carregar o dataset

```python
dados_sms = pd.read_csv(caminho_dataset)  # Lê o arquivo CSV e armazena os dados em um DataFrame
```

Verificar se carregou corretamente

```python
dados_sms.head()  # Exibe as cinco primeiras linhas do dataset
```

Descobrir o tamanho

```python
dados_sms.shape  # Retorna a quantidade de linhas e colunas do dataset
```

Ver todas as colunas

```python
dados_sms.columns.tolist()  # Converte os nomes das colunas para uma lista e exibe todos eles
```

Informações gerais

```python
dados_sms.info()  # Exibe quantidade de registros, valores não nulos e tipos de cada coluna
```

Procurar valores nulos

```python
dados_sms.isnull().sum()  # Conta quantos valores vazios existem em cada coluna
```

Verificar o alvo

```python
dados_sms["possivel_golpe"].value_counts()  # Conta quantas mensagens pertencem a cada classe
```

```python
dados_sms["possivel_golpe"].value_counts(normalize=True) * 100  # Calcula a porcentagem de cada classe
```

Verificar a carga emocional

```python
dados_sms["carga_emocional"].value_counts()  # Conta quantas mensagens existem em cada categoria de carga emocional
```
